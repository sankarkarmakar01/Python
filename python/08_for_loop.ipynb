WAP to enter 5 numbers from user append it into a list and print largest element from it. Without using max function

```python
l = []

for i in range(5):
    a = int(input("Enter a number: "))
    l.append(a)

print(l)

m = l[0]
for i in range(1, len(l)):
    if m < l[i]:
        m = l[i]

print("Maximum Element =", m)
```
