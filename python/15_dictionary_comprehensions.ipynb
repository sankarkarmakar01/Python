```python
# Normal
square_dict = {}

for i in range(1,6):
    square_dict[i] = i*i

print(square_dict)
```

```python
square_dict = {i: i*i for i in range(1,6)}
print(square_dict)
```

```python
numbers = range(1,11)
even_squares = {n: n*n for n in numbers if n % 2 == 0}
print(even_squares)
```

```python
original = {"a":1, "b":2, "c":3}
swapped = {v:k for k,v in original.items()}
print(swapped)
```

```python
numbers = range(1,6)
result = {n: "Even" if n % 2 == 0 else "Odd" for n in numbers}
print(result)
```

```python
celsius = [0,10,20,30]
fahrenheit = {c: (c*9/5)+32 for c in celsius}
print(fahrenheit)
```
