```python
def squre(num):
    for i in range(1,num + 1):
        yield i**2
```

```python
gen =squre(10)

print(next(gen))
print(next(gen))
print(next(gen))
```

```python
def mera_range(start, end) :
    for i in range(start,end):
        yield i
```

```python
for i in mera_range(15,26):
    print(i)
```
