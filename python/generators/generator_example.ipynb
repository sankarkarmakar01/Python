```python
import os
import cv2

def image_data_render(folder_path):

    for file in os.listdir(folder_path):
        f_array = cv2.imread(os.path.join(folder_path, file))
        yield f_array
```

```python
gen = image_data_render('D:\py_img')

next(gen)
next(gen)
next(gen)
```

## Chaining Generators

```python
def fibonacci_number(nums):
    x,y = 0,1
    for _ in range(nums):
        x,y = y , x +y
        yield x

def squre(nums):
    for num in nums:
        yield num**2

print(sum(squre(fibonacci_number(10))))
```
