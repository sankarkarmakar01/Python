```python
n = int(input("Enter a number"))
temp = n
s = 0

while n > 0:
    r = n % 10
    s = s * 10 + r
    n = n // 10

if s == temp:
    print("Palindrome")
else:
    print("Non Palindrome")
```

WAP to enter a number from user and check if it is armstrong or not

```python
n = int(input("Enter a number"))
temp = n
s = 0

while n > 0:
    r = n % 10
    s = s + (r*r*r)
    n = n // 10

if s == temp:
    print("Armstong")
else:
    print("Non Armstrong")
```

WAP to enter a number from user and print its individual digits in character format

```python
n = int(input("Enter a number"))
s = 0

while n > 0:
    r = n % 10
    s = s * 10 + r
    n = n // 10

while s > 0:
    r = s % 10
    if r == 1:
        print("One")
    elif r == 2:
        print("Two")
    elif r == 3:
        print("Three")
    elif r == 4:
        print("Four")
    elif r == 5:
        print("Five")
    elif r == 6:
        print("Six")
    s = s // 10
```

WAP to enter a limit from user and print fibonacii series

```python
n = int(input("Enter Limit"))
a = 0
b = 1

i = 1

while i <= n:
    print(a, end='  ')
    c = a + b
    a = b
    b = c
    i += 1
```
